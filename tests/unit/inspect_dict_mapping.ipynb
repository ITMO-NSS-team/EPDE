```python
import sys
sys.path.append('../..')

from functools import partial

import numpy as np
from epde.operators.utils.template import dictApplyUFunc
```

```python
features = {0: np.ones(shape=(5, 10)), 1: np.ones(shape=(5, 11))}
target = {0: np.ones(shape=(1, 10)), 1: np.ones(shape=(1, 11))}
```

```python
dictApplyUFunc(np.dot, features, dictApplyUFunc(np.transpose, target))
```

```python
a = {0: np.arange(-5, 5, 1), 1: np.arange(-6, 6, 1)}
```

```python
dictApplyUFunc(lambda x, y: np.where(x, 1, y), dictApplyUFunc(partial(np.isclose, b = 0), a), a)
```

```python
dictApplyUFunc(lambda x: x**2, a)
```
