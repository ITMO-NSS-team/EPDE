# Tutorial notebook

In this notebook we shall present and explain methods/functions from EPDE framework.

EpdeSearch constructor is used to create search object.

Explanation of EpdeSearch constructor parameters:

 - ``use_solver = False`` — skip the automatic PDE solver when scoring candidate equations; faster, less precise fitness estimate

 - argument ``multiobjective_mode `` select mode of multi-objective search: Pareto front by Discrepancy (quality) + Instability
 
 - parametr ``use_pic`` — define second pareto axis (Instability)
 - parametr ``coordinate_tensors`` describes grid coordinate tensors (from np.meshgrid) matching the data array
 - parametr ``device`` defines device ('cpu'/'cuda') for torch computations
 - argument ``boundary`` sets number of points to ignore in axes from each edge of the domain when computing derivatives
 - argument ``use_default_strategy`` sets evolutionary strategy( default strategy / custom user-defined(advanced feature))
 - argument ``director`` is advanced/internal optional parametr that assembles/coordinates the evolutionary strategy
 - argument `` director_params`` defines parameters forwarded to the strategy director, if one is built internally
 - argument ``time_axis`` sets index of the time axis in the data/coordinate grids
 - argument ``define_domain`` the problem domain is set up right away during initialization
 - argument ``function_form`` -  optional custom auxiliary function used in the weak derivative definition
 - argument ``verbose_params`` controls what algorithm details are printed/shown to the user during the search
 - argument ``memory_for_cache`` - rough memory budget for caching pre-evaluated tensors during the equation search
 - argument ``prune_domain`` - subdomains with no dynamics get pruned from the data
 - argument ``pivotal_tensor_label`` selects which token's values are used as the criterion for pruning
 - argument `` pruner`` - optional customer pruner object, that removes subdomains where the pivot derivative is identically zero
 - argument ``division_fractions`` - number of subdomains to split the domain into along each
 - argument ``discrepancy_metric`` is name of the discrepancy/error metric used as the primary fitness objective
 - argument ``solver_backend`` - backend used for solver-based fitness when use_solver=True (autograd / deepxde)


Methods ``set_preprocessor`` and ``set_moeadd_params`` set preprocessor settings, hyperparameters of EPDE algorithm

Here you can see explanation of ``set_preprocessor`` arguments meaning:
- argument ``preprocessor_pipeline`` is  a ready-made preprocessing pipeline object, created for a custom pipelines

- argument ``default_preprocessor_type``key selecting a built-in preprocessor: 'poly' - smoothing + differentiation
                                                                               'ANN' - neural-network data approximation
                                                                               'spectral' - Fourier-based differentiation
                                                                               'FD' - plain finite-difference differentiation
- argument ``preprocessor_kwargs``- extra keyword arguments forwarded to the chosen built-in preprocessor's builder method

Explanation of set_moeadd_params parameters:

- argument `population_size` sets the size of the solution population maintained during the multi-objective (MOEA/DD) optimization
- argument `solution_params` defines extra parameters forwarded into every newly created candidate solution object
- argument `H` sets the uniform-spacing parameter for the weight vectors (currently ignored — the optimizer always derives it as `population_size - 1`)
- argument `neighbors_number` sets the number of neighboring weight vectors considered as "neighbors" of a given sector during evolutionary operators
- argument `nds_method` selects the function used for non-dominated sorting of candidate solutions (default: NSGA-II fast non-dominated sorting)
- argument `ndl_update_method` selects the function used to insert a new solution point into the non-domination levels without recomputing them for the whole population
- argument `subregion_mating_limitation` defines the probability that mating selection is restricted to the subregions adjacent to a solution's own weight vector
- argument `PBI_penalty` sets the penalty parameter (theta) used in the Penalty-based Boundary Intersection (PBI) calculation
- argument `training_epochs` sets the maximum number of evolutionary optimization iterations (training stops early on Pareto-front convergence)
- argument `neighborhood_selector` selects the operator used to pick neighboring weight vectors (deprecated — ignored, kept for backward compatibility)
- argument `neighborhood_selector_params` defines parameters for `neighborhood_selector` (deprecated — ignored, kept for backward compatibility)
- argument `early_stopping_callback` defines an optional callback function checked during optimization to decide whether to stop training early

After that we shall explain arguments in fit method.
Here you can see the explanation of parametrs parametr in fit method:
- argument ``data`` desctibes array of target variable
- number of terms, accepted into the structure before regularization is regulated by argument ``equation_terms_max_number``
- argument ``equation_factors_max_number`` defines number of terms in the term can be set as integer
- argument ``variable_names`` defines names of the independent variable(s) being modeled.
- argument ``max_deriv_order`` sets the highest order of derivatives, that can be used in equation search

- argeument ``derives = None``  means that algorithm calculates derivatives in the method, skips pre-computed values 
- argument ``additional_tokens`` sets tokens, that would be used to construct the equations among the main variables and their derivatives
- argument ``data_fun_pow`` describes maximum power of token
- argument ``deriv_fun_pow`` sets max power to which derivative tokens can be raised
- argument ``optimizer``  - pre-built optimizer instance
- argument ``pool`` -  pre-built token pool, if you want to skip pool
- argument ``population`` -  pre-built population of candidate equations
- argument ``data_nn=None`` -  optional neural-network representation of the data
- argument ``ann_epochs_max`` sets Max number of training epochs for the NN preprocessing
- argument ``fourier_layer`` controls whether the data-representation ANN used for solver-based fitness
- argument ``fourier_params: dict = {'L': [4,], 'M': [3,]}`` - parameters for the Fourier layers above( 'L' — characteristic length scale(s), 
                                                                                                        'M' — number of Fourier modes/harmonics.)

The discovered equations can be accessed with ``EpdeSearch.equations()`` method. If the ``only_print = True``, than the equations will be printed in their text forms. 

Otherwise, they will be return: if the followup argument ``only_str`` is ``True``, than the equations are returned only in their symbolic string-form. If ``only_str = False``, than the equation in its program implementation ``epde.structure.main_structures.SoEq`` are returned.

Additional parametr ``num`` define number of poretto-levels for printing

We can use visual analysis of the Pareto-optimal set of candidate equations, placed in the complexity-quality criteria space, to select the optimal equation. Method ``EpdeSearch.visualize_solutions()`` illustrates the first non-dominated set of solutions (i.e. Pareto frontier).
